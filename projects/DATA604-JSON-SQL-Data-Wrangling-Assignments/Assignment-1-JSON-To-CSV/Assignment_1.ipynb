**Question 4 (5 marks)**

Using the json library again, read the contents of the file into a json object, and wrangle the object however you like, using Python built-ins, or by running the results through json library methods a second time. 

List the names of the Recipes that contain "Black pepper" as an ingredient.

```python
import json

with open("recipes.json", "r") as file:
    data = json.load(file)

pepper_recipes = set()

for category in data["Categories"]:

    # Book recipes + their components
    for book in category.get("Book", []):
        for recipe in book.get("Recipes", []):
            # Main recipe ingredients
            if "Black pepper" in recipe.get("Ingredients", []):
                pepper_recipes.add(recipe["Recipe"])
            # Component recipe ingredients
            for comp in recipe.get("Components", []):
                if "Black pepper" in comp.get("Ingredients", []):
                    pepper_recipes.add(comp["Recipe"])

    # Blog recipes
    for blog in category.get("Blog", []):
        for recipe in blog.get("Recipes", []):
            if "Black pepper" in recipe.get("Ingredients", []):
                pepper_recipes.add(recipe["Recipe"])

print("Recipes containing Black pepper:")
for r in pepper_recipes:
    print("-", r)
```

## Part C: Converting from JSON to CSV (13 marks)

Write your own code, using Python built-ins (i.e. no using pandas), the csv and json modules, to convert the data in the provided file into appropriate CSVs for each of your identified elements from Part B, Question 1. You may add additional fields where necessary. Make sure there are no fields with multiple values.

Each row of each of your CSVs should contain sufficient information so that they could be used in another program independently of each other. For example, you may be able to write another program which could read your CSV file about recipes to extract information about ingredients and preparation time (maybe even what book the recipe comes from).

```python
import json, csv

# Open the JSON file
with open("recipes.json", "r") as file:
    data = json.load(file)

# Categories CSV
with open("categories.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["CategoryName"])
    for category in data["Categories"]:
        writer.writerow([category["CategoryName"]])

# Books CSV 
with open("books.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["BookTitle", "ISBN", "CategoryName"])
    for category in data["Categories"]:
        for book in category.get("Book", []):
            writer.writerow([book["BookTitle"], book["ISBN"], category["CategoryName"]])

# Blogs CSV 
with open("blogs.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Title", "Author", "Address", "CategoryName"])
    for category in data["Categories"]:
        for blog in category.get("Blog", []):
            writer.writerow([blog["Title"], blog["Author"], blog["Address"], category["CategoryName"]])

# Recipes CSV 
with open("recipes.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Recipe", "PreparationTime", "CategoryName", "SourceType", "SourceTitle"])
    for category in data["Categories"]:
        for book in category.get("Book", []):
            for recipe in book.get("Recipes", []):
                writer.writerow([recipe["Recipe"], recipe["PreparationTime"],
                                 category["CategoryName"], "Book", book["BookTitle"]])
        for blog in category.get("Blog", []):
            for recipe in blog.get("Recipes", []):
                writer.writerow([recipe["Recipe"], recipe["PreparationTime"],
                                 category["CategoryName"], "Blog", blog["Title"]])

# Ingredients CSV 
with open("ingredients.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Recipe", "Ingredient"])
    for category in data["Categories"]:
        for book in category.get("Book", []):
            for recipe in book.get("Recipes", []):
                for ingredient in recipe.get("Ingredients", []):
                    writer.writerow([recipe["Recipe"], ingredient])
        for blog in category.get("Blog", []):
            for recipe in blog.get("Recipes", []):
                for ingredient in recipe.get("Ingredients", []):
                    writer.writerow([recipe["Recipe"], ingredient])
```
